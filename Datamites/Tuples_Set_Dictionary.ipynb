# Tuples

- Tuple is similar to a list except that a tuple is immutable.
- we cannot change the element of a tuple once assigned.
- when we do not want to change the data over tome, tuple is preferred.
- Iterating over the elements of a tuple is faster as compared to lists.

```python
measures = (2.3, 4.5, 5.6, 6.7)
type(measures)
```

```python
mixed = (2.3, 4.5, 3, "data") # mixed case tuple
type(mixed)
```

```python
mixed[3] # indexed calling in tuples
```

```python
mixed[0:2] # Slicing using index values similar to lists
```

```python
tup1 = ("Data", 25, [50,100], [150,90], ["Aparna", "Bella"], (99,22,23))
```

```python
tup1[4][1]
```

```python
tup1[5][1]
```
