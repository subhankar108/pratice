# If Statement

The if statement evaluates an expression and executes a block of code on the condition 
that the expression is true.

In case the condition is False, the 'if' block of code is ignored.

```python
raining_outside = True
if raining_outside:
    print ('Take Umbrella')
```

```python
raining_outside = 'yes'
if raining_outside == 'yes':
    print ('Take Umbrella')
```

```python
raining_outside = 'no' 
if raining_outside == 'yes':
    print ('Take Umbrella')
# It does not print because it checks the raining_outside condition with the above condition.
# Here it wasn't satisfied, so it won't print.
```

```python
age = 21
if age > 18:
    print('The person was eligiable to vote.')
    print ('The person is an adult.')
```

```python
age = 21
if age > 23:
    print('The person was eligiable to vote.')
    print ('The person is an adult.')
print('0')
```

```python
x = 20 
if x > 15:
    print('x is greater then 15')
if x < 18: # This line won't print because the condition is not satisfied.
    print('x is less then 22')
if x < 22:
    print('x is less then 22')
if x == 20:
    print('x is equal to 20')
```

# Else

Else may be included to account for alternate statement when the if condition is not true.

```python
x = 5
if x == 5:
    print('The if condition is true')
else: 
    print('The else block of code is executed')
```

```python
x = 5
if x > 0:
    print('The number is positive')
else:
    print('The number is negative')
```

# Nested if

An if condition can be said to be nested when it is a part of another if block of code. The nested if will only tested if 
the parent 'if' expression happens to be True.`

```python
i = 10
if i>=10:# The parent if condition
    #these expression are checked only if the parent if condition id True.
    if i<15:#nested if condition
        print('i is smaller then 15.')
    if i<12:
        print('i is smalle than 12 also')
else:
    print('x is greater then 10')
```

```python
from datetime import date, timedelta
day = int(input('Enter day: '))
month = int(input('Enter month: '))
year = int(input('Enter year: '))
days = int(input('Enter days: '))

birthday = date(year, month, day)
futur_date = birthday + timedelta(days)

age_in_days = (futur_date - birthday).days
age_in_years = age_in_days // 365

print('Date after {} days:'.format(days), futur_date)
print('Your age will be:{} years'.format(age_in_years))
```
