# Loops

- The process of performing a set of tasks iteratively as per requirement is known as looping.

- There are 2 essential loops: For and while.

## while loop

- while loops is used to execute a block of statements repeatedly subject to a condition still being true.

```python
i = 1 # Initialization
while i<6: # Condition
    print("while condition is True for:", i) # Block of code
    i=i+1 # Iterator
```

```python
# print even numbers between 0-10
c = 0 
while c<11:
    print(c)
    c=c+2 # Iterator
```

```python
# print odd numbers between 0-10
c = 1 
while c<11:
    print(c)
    c=c+2 # Iterator
```

```python
nums = [25, 40, 25, 80, 75, 25, 88, 25, 24, 25, 77, 25, 56, 25, 20, 25]
print(nums.count(25))
```

```python
while 25 in nums: # list membership check
    nums.remove(25) #perform removal of item from list
    c = nums.count(25) #checking the number of elements after removal
    if c == 5: # condition where while loop needs to break
        break # break will stops the loop immediately
print(nums) # print function after completion of the while loops
```

```python
i = 0
while i<5:
    i += 1 #(same as i=i+1)
    if i == 3:
        continue
    print(i)
```

# For loop

- For loops are used to loop through an iterable object, typically a list, tuple or set.

- The set of operations is performed on every item of the list.

```python
count = [10, 12, 2, 14, 25, 16, 22, 34, 45]
for c in count:
    print(c)
```

```python
fruit = ["Apple", "Banana", "Kiwi", "Orange"]
for i in fruit: # i acts as a placeholder for every item of the list untill it is exhausted
    print(i)
```

```python
s = "#12th main road"
for i in s:
    print(i)
```

### Range function

```python
for i in range(10):
    print(i)
```

```python
# For all odd numbers in between 0-20
for i in range(1,20,2): # (Start,Stop,Step)
    print(i)
```

```python
# For all even numbers in between 0-20
for i in range(0,20,2): # (Start,Stop,Step)
    print(i)
```

### nested for loops

```python
Colours = ["red", "blue", "green"]
Fruits = ["Apple", "Banana", "Orange"]
for i in Colours:
    for j in Fruits:
        print(i,j)
```

```python
course = ["math", "science", "social studies"]
batch_size = [78, 90, 56]
for i in course:
    for j in batch_size:
        print(i,j)
```

# Functions

- A function is a block of organised code written to carry out a specified task.

- A function does not necessarily return data.

- Functions help to break the program into smaller modular chunks for better readability.

- Parameters are specified after the function name inside the parentheses().

- Functions start using the 'def' keyword.

```python
def area_of_circle(r): # def-start function, area_of_circle-name of the function, (r)-parameter:
    a = 3.14*r*r
    print(a)
```

```python
area_of_circle(2) # Argument
```

```python
def perimeter_of_square (side):
    peri = 4*side
    print(peri)
```

```python
perimeter_of_square(5)
```

```python
def area_of_rectangle(l,b):
    area= l*b
    print(area)
```

```python
area_of_rectangle(2,3)
```

```python
def details(name, userid, country):
    print("Name:-", name)
    print("userid:-", userid)
    print("country:-", country)
```

```python
details('Bapu','220301120106','India')
```

```python
def even_or_odd():
    a = eval(input("Enter a number:"))
    if a % 2 == 0:
        print('The number is even.')
    else:
        print('The number is odd.')
```

```python
even_or_odd()
```
