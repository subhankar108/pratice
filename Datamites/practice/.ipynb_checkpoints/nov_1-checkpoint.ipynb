## Ask the user to enter three numbers and find the largest without using max().

```python
num1=eval(input('enter a number'))
num2=eval(input('enter a number'))
num3=eval(input('enter a number'))

if num1>num2 and num1>num3:
    print(num1)
elif num2>num1 and num2>num3:
    print(num2) 
else:
    print(num3)
```

## Ask the user to enter a year and check whether it is a leap year.

```python
year_num=eval(input('enter the year:'))

if year_num%400==0:
    print('It was a leap year.')
elif year_num%100==0:
    print('Not a leap year.')
elif year_num%4==0:
    print('It was a leap year.')
else:
    print('Not a leap year.')
```

## Create a calculator using if-elif-else that accepts two numbers and an operator (+, -, *, /)

```python
num1=eval(input('enter a number'))
operator=input('enter operator')
num2=eval(input('enter a number'))

if operator=='+':
    print(num1+num2)
elif operator=='-':
    print(num1-num2)
elif operator=='*':
    print(num1*num2)
elif operator=='/':
    if num2!=0:
        print(num1/num2)
    else:
        print('Cannot divide by zero')
else:
    print('Invalid operator')# for other operator like %,// etc..
```

## Ask the user to enter a number and:
1. Check whether it is positive, negative or zero.
2. If positive, check whether it is even or odd.

```python
num=eval(input('enter a number:'))

if num>0:
    print(f'{num} is positive.')
    if num%2==0:
        print(f'{num} is even.')
    else:
        print(f'{num} is odd.')
elif num<0:
    print(f'{num} is negative.')
else:
    print('number is zero.')
```

## Ask the user to enter their marks in five subjects. Calculate the percentage and display the grade.

```python
math=eval(input('enter mark of math:'))
bio=eval(input('enter mark of bio:'))
phy=eval(input('enter mark of phy:'))
chem=eval(input('enter mark of chem:'))
eng=eval(input('enter mark of eng:'))

if 0>math<100:
    print('invalid mark')
elif 0>bio<100:
    print('invalid mark')
elif 0>phy<100:
    print('invalid mark')
elif 0>chem<100:
    print('invalid mark')
elif 0>eng<100:
    print('invalid mark')
else:   
    total_mark=math+bio+phy+chem+eng
    percentage=total_mark/5
    if percentage>=90:
        print('O')
    elif percentage>=80:
        print('E')
    elif percentage>=60:
            print('A')
    elif percentage>=40:
        print('Below avrage')
    else:
        print('fail')
```
