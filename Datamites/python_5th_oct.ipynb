## Typecasting 

Typecasting is a process of a variable's data type converted into another type in order for an operation to be preformed.

### Auto Typecasting

The python automatically one data type to another without any user involvement.

```python
4 + 5.67 + False # Background processing of the code changes boolean value of false to 0.
```

```python
4 + 5.67 + True # Here True to 1.
```

```python
4 + 5.67 + 'Data' # String typecasting is not possible here.
```
