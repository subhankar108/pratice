```python
i=1
while True:
    user= random.randint(1,10)
    user2=eval(input('enter number'))
    if user==user2:
        print('won')
    else:
        print('lose')
    i=i+1
    if i==4:
        break
```

```python
num = eval(input(' enter a number'))

total=0
while num>0:
    reminder=num%10
    total=total+reminder
    num=num//10
print(total)
```

```python
num = eval(input(' enter a number'))

reverse=0
while num>0:
    remain=num%10
    reverse = reverse*10+remain
    num=num//10
print(reverse)
```
